```python
import os
import json
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# estilo base
sns.set_theme(style="white")

plt.rcParams.update({
    "font.family": "serif",
    "font.size": 14,
    "axes.labelsize": 12,
    "axes.titlesize": 14,
    "legend.fontsize": 12,
    "xtick.labelsize": 12,
    "ytick.labelsize": 12,
    "lines.linewidth": 1,
    "axes.linewidth": 0.8,
    "figure.dpi": 300
})

# plt.rcParams.update({
#     "font.size": 14,          # tamanho base
#     "axes.titlesize": 16,     # título dos plots
#     "axes.labelsize": 14,     # labels dos eixos
#     "xtick.labelsize": 12,    # números eixo X
#     "ytick.labelsize": 12,    # números eixo Y
#     "legend.fontsize": 12     # legenda
# })

# 1 coluna
FIGSIZE_SINGLE = (7, 5)

# 2 colunas
FIGSIZE_DOUBLE = (7, 3)

#palette = sns.color_palette("colorblind")
palette = sns.color_palette("Set2") #("flare") #("Set2") #("husl", 9) #("pastel")

# pasta com os ficheiros
DATA_DIR = "./"

data = []

def parse_filename(filename):
    # remove prefixo e extensão
    name = filename.replace("results_", "").replace(".json", "")
    
    parts = name.split("_")
    
    attack = parts[0]
    type = parts[1]  # adversarial, random, etc.
    feature = None
    ratio = 1
    if len(parts) > 2:
        feature = parts[2]  # size, iat, both
        
        if len(parts) > 3:
            # extrair ratio
            ratio_part = parts[3]
            if "rest" in ratio_part:
                ratio = float(ratio_part.replace("rest", "")) / 10

    return attack, type, feature, ratio


# ler todos os ficheiros
# iterar sobre as pastas na pasta
for folder in os.listdir(DATA_DIR):
    folder_path = os.path.join(DATA_DIR, folder)
    if os.path.isdir(folder_path):
        for folder2 in os.listdir(folder_path):
            folder2_path = os.path.join(folder_path, folder2)
            if os.path.isdir(folder2_path):
                for file in os.listdir(folder2_path):
                    if file.endswith(".json"):
                        with open(os.path.join(folder2_path, file)) as f:
                            content = json.load(f)
                    
                            attack, type, feature, ratio = parse_filename(file)

                            if type == "original":
                                data.append({
                                    "attack": attack,
                                    "feature": type,
                                    "ratio": 0,
                                    "model": content["model"],
                                    "ASR": content["metrics"]["attack_success_rate"],
                                    "F1": content["metrics"]["f1_score"]
                                })
                            else:
                                data.append({
                                    "attack": attack,
                                    "feature": feature,
                                    "ratio": ratio,
                                    "model": content["model"],
                                    "ASR": content["metrics"]["attack_success_rate"],
                                    "TrainLoss": content["Train_loss"]
                                })


df = pd.DataFrame(data)

# garantir ordem consistente
feature_order = ["original", "iat", "size", "both"]
df["feature"] = pd.Categorical(df["feature"], categories=feature_order, ordered=True)

df["feature"] = df["feature"].map({
    "original": "Original",
    "iat": "IAT",
    "size": "Size",
    "both": "Both"
})

palette_fixed = {
    "Original": palette[0],
    "IAT": palette[1],
    "Size": palette[2],
    "Both": palette[3],
}

# garantir ordem consistente
attack_order = ["BruteForce", "DDoS", "PortScan"]
df["attack"] = pd.Categorical(df["attack"], categories=attack_order, ordered=True)

df["attack"] = df["attack"].map({
    "BruteForce": "Brute Force",
    "DDoS": "DDoS",
    "PortScan": "Port Scan"
})

ratio_order = [0, 0.2, 0.5, 1]
df["ratio"] = pd.Categorical(df["ratio"], categories=ratio_order, ordered=True)
# converter ratio para percentagem se existir
df["ratio"] = df["ratio"].apply(lambda x: f"{int(x*100)}%" if pd.notnull(x) else x)

df["TrainLoss"] = df["TrainLoss"] + 4

models_order = ["CNN", "LR", "MLP", "RF", "SVM"]
df["model"] = pd.Categorical(df["model"], categories=models_order, ordered=True)
```

```python
for attack in df["attack"].unique():
    # filtrar dados para o ataque atual
    subset = df[df["attack"] == attack].copy()

    #filtrar apenas os dados com ratio 1
    subset = subset[subset["ratio"] == "100%"]

    plt.figure(figsize=FIGSIZE_SINGLE)

    hue_order = ["IAT", "Size", "Both"]
    
    ax = sns.barplot(
        data=subset,
        x="model",
        y="ASR",
        hue="feature",
        hue_order=hue_order,
        palette=palette_fixed,
        errorbar=None
    )

    ax.set_xlabel("Model")
    ax.set_ylabel("ASR")

    for container in ax.containers:
        ax.bar_label(container, fmt="%.2f", fontsize=6, padding=2)

    # remover top/right spines
    sns.despine()

    # legenda compacta
    ax.legend(title="", frameon=False, loc="upper left", bbox_to_anchor=(1, 1))

    plt.tight_layout()
    plt.savefig(f"plots/ieee_asr_feature_{attack}.png", bbox_inches="tight", dpi=300)
    plt.close()
```

```python
from matplotlib import colors
import numpy as np

xmin = subset["TrainLoss"].min()
xmax = subset["TrainLoss"].max()
ymin = subset["ASR"].min()
ymax = subset["ASR"].max()

extent = [xmin, xmax, ymin, ymax]

plt.figure(figsize=FIGSIZE_SINGLE)

subset1 = df[df["ratio"] == "100%"]
subset = subset1[subset1["feature"].isin(["IAT", "Size"])]

# separar
iat = subset[subset["feature"] == "IAT"]
size = subset[subset["feature"] == "Size"]

norm = colors.Normalize(vmin=1, vmax=5)

def truncate_colormap(cmap, minval=0.3, maxval=1.0, n=100):
    return plt.cm.colors.LinearSegmentedColormap.from_list(
        f"trunc({cmap.name})",
        cmap(np.linspace(minval, maxval, n))
    )

cmap_iat = truncate_colormap(plt.cm.YlGnBu, 0.3, 1.0)
cmap_size = truncate_colormap(plt.cm.Reds, 0.3, 1.0)

hb1 = plt.hexbin(
    iat["TrainLoss"], iat["ASR"],
    gridsize=30,
    extent=extent,
    cmap=cmap_iat,
    norm=norm,
    mincnt=1,
    alpha=0.8
)

hb2 = plt.hexbin(
    size["TrainLoss"], size["ASR"],
    gridsize=30,
    extent=extent,
    cmap=cmap_size,
    norm=norm,
    mincnt=1,
    alpha=0.8
)

cb = plt.colorbar(hb1)
cb.set_label("IAT Count", loc="bottom")

cb = plt.colorbar(hb2)
cb.set_label("Size Count", loc="bottom")


plt.tight_layout()
plt.savefig("plots/ieee_asr_vs_loss.png", bbox_inches="tight", dpi=300)
plt.close()
```

```python
from scipy.stats import pearsonr, spearmanr

for feature in ["IAT", "Size"]:
    sub = subset[subset["feature"] == feature]
    
    pearson_corr, p_p = pearsonr(sub["TrainLoss"], sub["ASR"])
    spearman_corr, p_s = spearmanr(sub["TrainLoss"], sub["ASR"])
    
    print(f"\nFeature: {feature}")
    print(f"Pearson: {pearson_corr:.3f} (p={p_p:.4f})")
    print(f"Spearman: {spearman_corr:.3f} (p={p_s:.4f})")

import statsmodels.api as sm

for feature in ["IAT", "Size"]:
    sub = subset[subset["feature"] == feature]
    
    X = sub["TrainLoss"]
    y = sub["ASR"]
    
    X = sm.add_constant(X)
    model = sm.OLS(y, X).fit()
    
    print(f"\nFeature: {feature}")
    print(model.summary())

threshold = subset["TrainLoss"].median()

iat = subset[subset["feature"] == "IAT"]

low_loss = iat[iat["TrainLoss"] <= threshold]["ASR"]
high_loss = iat[iat["TrainLoss"] > threshold]["ASR"]

from scipy.stats import ttest_ind

t_stat, p_val = ttest_ind(low_loss, high_loss)

print("T-test p-value:", p_val)

# Conclusão do chat: Nada a ver
```

```python
def build_matrix(attacks, models):
    matrix = np.zeros((len(models), len(attacks)))
    for j, attack in enumerate(attacks):
        # verify if there are any rows for this attack and model with feature "Original"
        if not ((df["attack"] == attack) & (df["feature"] == "Original")).any():
            print(f"No data for attack {attack} with feature Original")
            continue
        for i, model in enumerate(models):
            if not ((df["attack"] == attack) & (df["model"] == model) & (df["feature"] == "Original")).any():
                print(f"No data for attack {attack}, model {model} with feature Original")
                matrix[i, j] = np.nan
            else:
                matrix[i, j] = df[(df["attack"] == attack) & (df["model"] == model) & (df["feature"] == "Original")]["F1"].values[0]

    return matrix

# =========================
# PLOTTING
# =========================

def plot_heatmap(matrix, attacks, models, title, ax):
    sns.heatmap(
        matrix,
        ax=ax,
        annot=True,
        fmt=".3f",
        cmap="GnBu",
        vmin=0.0,
        vmax=1.0,
        xticklabels=attacks,
        yticklabels=models,
        linewidths=0.5,
        linecolor="white",
        cbar_kws={"label": "F1 Score"},
    )
    ax.set_title(title, fontsize=12, pad=10)
    ax.set_xlabel("Attack", fontsize=10)
    ax.set_ylabel("Model", fontsize=10)
    ax.tick_params(axis="x", rotation=0)
    ax.tick_params(axis="y", rotation=0)

# =========================
# MAIN
# =========================

attacks = df["attack"].cat.categories
models = df["model"].cat.categories

print("Attacks:", attacks)
print("Models:", models)

# --- CIC-IDS2017 ---
cicids_matrix = build_matrix(attacks, models)

fig, ax = plt.subplots(figsize=(6, 4))
plot_heatmap(cicids_matrix, attacks, models,
             "Baseline Detection Performance — CIC-IDS2017", ax)
plt.tight_layout()
plt.savefig("plots/baseline_cicids.png", bbox_inches="tight")
plt.close()
print("Saved baseline_cicids")
```
